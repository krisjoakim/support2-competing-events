```python
import pickle
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

PROCESSED_DATA_DIR = PROJECT_ROOT / 'data' / 'processed'
ARTIFACTS_DIR = PROJECT_ROOT / 'artifacts'
RESULTS_DIR = PROJECT_ROOT / 'results'
```

# Notebook 3: Prediction generation

This historical notebook generates predictions for the course-provided sample. The 2026 revisit only updates repository-relative paths and publication context.

Laster inn modellen:

```python
with (ARTIFACTS_DIR / 'model.pkl').open('rb') as model_file:
    model = pickle.load(model_file)
final_sample_df = pd.read_csv(PROCESSED_DATA_DIR / 'final_sample_df.csv')
```

```python
df = final_sample_df.drop(columns="pasient_id")
```

Predikerer og laster prediksjonene opp som csv-fil:

```python
predictions = model.predict(df)
predictions_df = pd.DataFrame(predictions, columns=['predictions'])
predictions_df['pasient_id'] = final_sample_df['pasient_id'].values
predictions_df = predictions_df[['pasient_id', 'predictions']]

RESULTS_DIR.mkdir(parents=True, exist_ok=True)
predictions_df.to_csv(RESULTS_DIR / 'predictions.csv', index=False)
```
